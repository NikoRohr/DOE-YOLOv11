```python
from PIL import Image
import os
import matplotlib.pyplot as plt

DATA_DIR = "/Users/michelle/Documents/Masterstudium/WiSe25_26/Computational Imaging/Project/data/"
```

```python
def image_grid(img_dir, n, rows, cols, figsize=(8, 4.5)):

    # Get n sample file names
    filenames = os.listdir(img_dir)[:n]
    images = [Image.open(os.path.join(img_dir, f)) for f in filenames]

    # Create a grid
    plt.figure(figsize=figsize)

    for i, img in enumerate(images):
        plt.subplot(rows, cols, i + 1)
        plt.imshow(img)
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
image_grid(IMAGE_DIR, 25, 5, 5)
```

```python
IMAGE_DIR = DATA_DIR + "DAWN/own_train_val_split/train/images"
image_grid(IMAGE_DIR, 25, 5, 5)
```
