```python
from PIL import Image
import os
import matplotlib.pyplot as plt
import random

DATA_DIR = "/Users/michelle/Documents/Masterstudium/WiSe25_26/Computational Imaging/Project/data/"
```

```python
def image_grid(img_dir, n, rows, cols, target_size=(224,224), figsize=(8, 4.5)):

    # Get n sample file names
    filenames = os.listdir(img_dir)
    sample_files = random.sample(filenames, n)
    images = [Image.open(os.path.join(img_dir, f)).resize(target_size) for f in sample_files]

    # Create a grid
    plt.figure(figsize=figsize)

    for i, img in enumerate(images):
        plt.subplot(rows, cols, i + 1)
        plt.imshow(img)
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
IMAGE_DIR = DATA_DIR + "foggy_cityscape/datasets/khitdon/foggy-cityscapes-yolo/versions/1/YOLO_dataset_foggycityscapes/train/images"
image_grid(IMAGE_DIR, 25, 5, 5, target_size=(2048,1024), figsize=(8, 4.5))
```

```python
IMAGE_DIR = DATA_DIR + "DAWN/own_train_val_split/train/images"
image_grid(IMAGE_DIR, 25, 5, 5, target_size=(2048,1024), figsize=(8, 4.5))
```
