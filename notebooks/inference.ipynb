```python
from ultralytics import YOLO
from PIL import Image
import os
```

```python
DATA_DIR = os.path.join(os.path.dirname(os.getcwd()), "data")
# iterate through dirs to find data.yaml. Really buggy way to do this
while True:
    # print("Searching in:", DATA_DIR)
    if "data.yaml" in os.listdir(DATA_DIR):
        break
    elif os.listdir(DATA_DIR) == [] or DATA_DIR == "/":
        raise FileNotFoundError("data.yaml not found in any parent directories")
    DATA_DIR = os.path.join(DATA_DIR, os.listdir(DATA_DIR)[-1])

print("DATA_DIR:", DATA_DIR)
```

```python
sample = os.listdir(DATA_DIR + "/train/images")[0]
image = Image.open(DATA_DIR + "/train/images/" + sample)
display(image)
```

```python
model = YOLO('yolo11n.pt')
result = model.predict(image, conf=0.25)[0]
```

```python
import matplotlib.pyplot as plt
plt.imshow(result.plot())
plt.show()
```
