```python
from ultralytics import YOLO
from PIL import Image
import os
import torch
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
DATA_DIR = os.path.join(os.path.dirname(os.getcwd()), "data")
# iterate through dirs to find data.yaml. Really buggy way to do this
while True:
    # print("Searching in:", DATA_DIR)
    if "data.yaml" in os.listdir(DATA_DIR):
        break
    elif os.listdir(DATA_DIR) == [] or DATA_DIR == "/":
        raise FileNotFoundError("data.yaml not found in any parent directories")
    DATA_DIR = os.path.join(DATA_DIR, os.listdir(DATA_DIR)[0])

print("DATA_DIR:", DATA_DIR)
```

```python
configs = {
    "data": DATA_DIR + "/data.yaml",
    "epochs": 10,
    "batch": 32,
    "imgsz": 640,
    "cache": True,
}
```
